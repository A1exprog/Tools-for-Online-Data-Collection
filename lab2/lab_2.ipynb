```python
import requests
import json
from lxml import html
```

```python
url = "https://books.toscrape.com/"
headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
    }
res = requests.get(url)
print(res.status_code)
tree = html.fromstring(res.content) # DOM-дерево
products = tree.xpath("//article[@class='product_pod']") # поиск блоков товаров
scraped_data = []
    
    # Извлекаем данные из каждой карточки
for product in products:
        # Название товара (атрибут title ссылки внутри тега <h3>)
    title_list = product.xpath(".//h3/a/@title")
    title = title_list[0] if title_list else "Название не найдено"
        
        # Ссылка на изображение (атрибут src тега <img> внутри блока с классом image_container)
    img_src_list = product.xpath(".//div[@class='image_container']/a/img/@src")
    img_src = img_src_list[0] if img_src_list else "Изображение не найдено"
    img_src = "https://books.toscrape.com/" + img_src.replace("../", "")
        
        # Цена товара (текст внутри тега <p> с классом price_color)
    price_list = product.xpath(".//p[@class='price_color']/text()")
    price = price_list[0] if price_list else "Цена не найдена"
        
       
    product_info = {
            "title": title,
            "image_url": img_src,
            "price": price
        }
    scraped_data.append(product_info)
        
    print(f"{title} - {img_src} - {price}")

    with open("books_data.json", "w", encoding="utf-8") as f:
        json.dump(scraped_data, f, ensure_ascii=False, indent=4)
```

```python
import pandas as pd

df = pd.DataFrame({'Title':[],'image_url':[], 'price':[]})
for i in range(len(scraped_data)):
    for key in scraped_data[i]:
        df.loc[i, "Title"] = scraped_data[i]['title']
        df.loc[i, "image_url"] = scraped_data[i]['image_url']
        df.loc[i, "price"] = scraped_data[i]['price']

df.to_csv('parsing_book.csv')
```

```python
import requests
from bs4 import BeautifulSoup
import json
url = "https://books.toscrape.com/"
headers = {
        "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/115.0.0.0 Safari/537.36"
    }
```

```python

res = requests.get(url, headers=headers)
res.raise_for_status()

soup = BeautifulSoup(res.content, "html.parser")

products = soup.find_all("article", class_="product_pod")

scraped_data = []

for product in products:

    title = product.find("h3").find("a").get("title")

    img_src = product.find("div", class_="image_container").find("img").get("src")

    if img_src.startswith("../"):
        img_src = url + img_src.replace("../", "")

    price = product.find("p", class_="price_color").get_text(strip=True)

    product_info = {
        "title": title,
        "image_url": img_src,
        "price": price
    }

    scraped_data.append(product_info)

    print(f"{title} - {img_src} - {price}")

with open("books_data_bs.json", "w", encoding="utf-8") as f:
    json.dump(scraped_data, f, ensure_ascii=False, indent=4)
```
